```python
# Q1 to Q4

import numpy as np

def element_matrix(coordinates):

    x = coordinates[0, :]
    y = coordinates[1, :]

    S1 = np.array([x[2]-x[1], y[2]-y[1]])
    S2 = np.array([x[0]-x[2], y[0]-y[2]])
    S3 = np.array([x[1]-x[0], y[1]-y[0]])


    A_tot = 0.5 * ((x[1]-x[0])*(y[2]-y[0]) - (x[2]-x[0])*(y[1]-y[0]))


    if A_tot < 0:
        raise ValueError("Nodes are not in counterclockwise order")

    if A_tot == 0:
        raise ValueError("Triangle has zero area")


# Q5 -Gradients of the basis functions (𝜑𝑖)

    grad_phi = np.array([[-S1[1], -S2[1], -S3[1]], [ S1[0],  S2[0],  S3[0]]]) / (2 * A_tot)


    #Q7
    Aij = np.zeros((3, 3))

    #Q8
    Aij = A_tot * np.dot(grad_phi.T, grad_phi)

    return Aij,A_tot

coordinates = np.array([[-0.5, 0.0, 0.6], [ 0.5, -0.2, 0.4]])

Aij,A_tot = element_matrix(coordinates)

print("Element Matrix Aij:")
print(Aij)

print("\nArea of Triangle:")
print(A_tot)


```
